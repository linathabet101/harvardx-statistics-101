### Example 1.5.2: Vandermonde's identity.

A famous relationship between binomial coefficients, called *Vandermonde's identity*, says that

$$
\binom{m + n}{k} = \sum_{j=0}^{k} \binom{m}{j} \binom{n}{k - j}
$$

This identity will come up several times in this course. Trying to prove it with a brute force expansion of all the binomial coefficients would be a nightmare. But a story proves the result elegantly and makes it clear *why* the identity holds.

#### Story proof

Consider a group of $m$ peacocks and $n$ toucans, from which a set of size $k$ birds will be chosen. There are $\binom{m + n}{k}$ possibilities for this set of birds. If there are $j$ peacocks in the set, then there must be $k - j$ toucans in the set. The right-hand side of Vandermonde's identity sums up the cases for $j$.

---

### Understanding the counting argument

We count the same group of $k$ birds in two different ways.

**Method 1 — Choose all birds together (left-hand side)**

- There are $m+n$ birds in total.
- Choose $k$ birds from all of them.

The number of possibilities is:

$$
\binom{m+n}{k}
$$

**Method 2 — Separate the choices by bird type (right-hand side)**

Suppose exactly $j$ peacocks are selected.

- Choose $j$ peacocks from the $m$ available: $\binom{m}{j}$.
- Choose the remaining $k-j$ toucans from the $n$ available: $\binom{n}{k-j}$.
- Multiply the two counts to obtain the number of groups containing exactly $j$ peacocks and $k-j$ toucans.

The number of such groups is:

$$
\binom{m}{j}\binom{n}{k-j}
$$

The number of peacocks selected can vary. We therefore consider every possible value of $j$ and add the corresponding counts:

$$
\sum_{j=0}^{k}\binom{m}{j}\binom{n}{k-j}
$$

Both methods count exactly the same groups of $k$ birds. Therefore,

$$
\boxed{
\binom{m+n}{k}
=
\sum_{j=0}^{k}\binom{m}{j}\binom{n}{k-j}
}
$$

**Important:** A binomial coefficient is zero when its lower argument is greater than its upper argument. Thus, terms corresponding to impossible selections automatically contribute zero.

```python
from math import comb

# Example: 4 peacocks, 3 toucans, choose 3 birds
m = 4
n = 3
k = 3

if not (
    isinstance(m, int) and isinstance(n, int) and isinstance(k, int)
    and m >= 0 and n >= 0 and 0 <= k <= m + n
):
    raise ValueError("Use non-negative integers m and n, with 0 <= k <= m+n.")

print("VANDERMONDE'S IDENTITY — VERIFICATION")
print("=" * 55)
print()
print("QUESTION")
print(f"From {m} peacocks and {n} toucans, how many different")
print(f"groups of {k} birds can we choose?")
print()
print("We will count the groups in two different ways.")
print()

# Left-hand side: choose k birds from all m+n birds
left = comb(m + n, k)

print("METHOD 1: Choose all birds together")
print(f"  C({m+n}, {k}) = {left}")
print()

# Right-hand side: sum over the possible numbers of peacocks
print("METHOD 2: Separate the choices by bird type")
print()

right = 0

for j in range(k + 1):
    peacocks = comb(m, j) if j <= m else 0
    toucans = comb(n, k - j) if 0 <= k - j <= n else 0
    term = peacocks * toucans
    right += term

    print(f"  j = {j}:")
    print(f"    Choose {j} peacocks: C({m}, {j}) = {peacocks}")
    print(f"    Choose {k-j} toucans: C({n}, {k-j}) = {toucans}")
    print(f"    Number of groups: {peacocks} × {toucans} = {term}")
    print()

print("CONCLUSION")
print(f"Left-hand side  = {left}")
print(f"Right-hand side = {right}")
print(f"Identity verified: {left == right}")
```

```python
print("VANDERMONDE'S IDENTITY — COUNTING BY CASES")
print("=" * 58)
print()

print("We have m peacocks and n toucans.")
print("We want to select k birds.")
print()
print("The number of selected peacocks can be:")
print()

for j in range(k + 1):
    peacocks = comb(m, j) if j <= m else 0
    toucans = comb(n, k - j) if 0 <= k - j <= n else 0
    total = peacocks * toucans

    print(f"  Case j = {j}:")
    print(f"    {j} peacocks + {k-j} toucans")
    print(f"    {peacocks} choices × {toucans} choices = {total} groups")
    print()

print("-" * 58)
print(f"Sum of all cases = {right}")
print(f"Choose {k} birds from all {m+n} birds = {left}")
print()
print("Every group belongs to exactly one case,")
print("according to the number of peacocks it contains.")
print("Therefore, adding all cases counts every group exactly once.")
```

### A quick note about the birds!

A small fun fact: *peacock* refers specifically to a male peafowl (females are called peahens, and young birds are peachicks!).

Peacocks and toucans aren't exactly neighbors! Indian peafowl are native to South Asia, while toucans live in Central and South America. So, seeing them hanging out together in the wild is highly unlikely (unless someone brings them together, for example, in a zoo!)
