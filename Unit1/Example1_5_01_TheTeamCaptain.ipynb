### Example 1.5.1: The team captain.

For any positive integers $n$ and $k$ with $k \le n$,

$$n \binom{n - 1}{k - 1} = k \binom{n}{k}$$

This is again easy to check algebraically, using the fact that $m! = m(m - 1)!$ for any positive integer $m$, but a story proof is more insightful.

#### Story proof

Consider a group of $n$ people, from which a team of $k$ will be chosen, one of whom will be the team captain. To specify a possibility, we could first choose the team captain and then choose the remaining $k - 1$ team members; this gives the left-side. Equivalently, we could first choose the $k$ team members and then choose one of them to be captain; this gives the right-hand side.

---

### Understanding the two counting methods

We count the same outcome in two different ways: a team of $k$ people together with one designated captain.

**Method 1 — Choose the captain first (left-hand side)**

- Choose the captain from all $n$ people: $n$ choices.
- Choose the other $k-1$ team members from the remaining $n-1$ people: $\binom{n-1}{k-1}$ choices.

Total: $n\binom{n-1}{k-1}$.

**Method 2 — Choose the team first (right-hand side)**

- Choose the $k$ team members from all $n$ people: $\binom{n}{k}$ choices.
- Choose the captain from those $k$ team members: $k$ choices.

Total: $k\binom{n}{k}$.

Both methods count every team-with-captain possibility exactly once. Therefore, the two expressions are equal.

```python
from math import comb

# Example: a group of 5 people and a team of 3
n = 5
k = 3

if not (isinstance(n, int) and isinstance(k, int) and 1 <= k <= n):
    raise ValueError("Use positive integers n and k with 1 <= k <= n.")

print("THE TEAM CAPTAIN — VERIFYING THE IDENTITY")
print("=" * 55)
print()
print("QUESTION")
print(f"From a group of {n} people, how many different ways can we")
print(f"choose a team of {k} people and designate one member as captain?")
print()
print("We will count the possibilities in two different ways.")
print()

# Method 1: choose the captain, then the other k-1 team members
left = n * comb(n - 1, k - 1)

# Method 2: choose the team, then its captain
right = k * comb(n, k)

print("METHOD 1: Captain first")
print(f"  Choose the captain: {n} choices")
print(f"  Choose the other {k-1} members: C({n-1}, {k-1}) = {comb(n-1, k-1)}")
print(f"  Total: {n} × {comb(n-1, k-1)} = {left}")
print()

print("METHOD 2: Team first")
print(f"  Choose the team: C({n}, {k}) = {comb(n, k)}")
print(f"  Choose its captain: {k} choices")
print(f"  Total: {k} × {comb(n, k)} = {right}")
print()

print("CONCLUSION")
print(f"Left-hand side  = {left}")
print(f"Right-hand side = {right}")
print(f"Identity verified: {left == right}")
```

```python
print("TWO WAYS TO COUNT A TEAM WITH A CAPTAIN")
print("=" * 52)
print()

print("METHOD 1 — CAPTAIN FIRST")
print("  Choose 1 captain from n people")
print("                 |")
print("                 v")
print("  Choose k-1 other members from the n-1 remaining people")
print("                 |")
print("                 v")
print("  n × C(n-1, k-1) possibilities")
print()

print("METHOD 2 — TEAM FIRST")
print("  Choose k team members from n people")
print("                 |")
print("                 v")
print("  Choose 1 captain from those k team members")
print("                 |")
print("                 v")
print("  C(n, k) × k possibilities")
print()

print("Both methods describe exactly the same final object:")
print("a selected team of k people, with one of its members marked")
print("as captain. Therefore the two counts must be equal.")
```

```python
# EXPERIMENT: Change n and k to explore the identity.
# Try, for example, n = 8 and k = 2, or n = 10 and k = 5.

n = 8
k = 2

if not (isinstance(n, int) and isinstance(k, int) and 1 <= k <= n):
    print("Choose positive integers n and k satisfying 1 <= k <= n.")
else:
    captain_first = n * comb(n - 1, k - 1)
    team_first = k * comb(n, k)

    print("TEAM CAPTAIN IDENTITY — YOUR EXPERIMENT")
    print("=" * 48)
    print(f"n = {n} people; k = {k} people on the team")
    print()

    print("Captain-first count:")
    print(f"  {n} × C({n-1}, {k-1})")
    print(f"  = {n} × {comb(n-1, k-1)}")
    print(f"  = {captain_first:,}")
    print()

    print("Team-first count:")
    print(f"  {k} × C({n}, {k})")
    print(f"  = {k} × {comb(n, k)}")
    print(f"  = {team_first:,}")
    print()

    print(f"Both counts are equal: {captain_first == team_first}")
```
